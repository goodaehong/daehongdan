```python
import os
import shutil
import zipfile

bundle = '/content/smoke_clean_dfire5_bundle'
zip_path = '/content/smoke_clean_dfire5_bundle.zip'
if os.path.exists(bundle):
    shutil.rmtree(bundle)
os.makedirs(bundle, exist_ok=True)
with zipfile.ZipFile(zip_path) as archive:
    archive.extractall(bundle)
required = [
    f'{bundle}/base_dfire_best.pt',
    f'{bundle}/train_colab.py',
    f'{bundle}/dataset/data_stage_a.yaml',
    f'{bundle}/dataset/data_final.yaml',
]
for path in required:
    print(os.path.exists(path), path)
assert all(os.path.exists(path) for path in required)
print('학습 파일 준비 완료')
```
